```python
from pathlib import Path

candidates = [
    Path("assets/fieldreport-page-screenshot.png"),
    Path("notebooks/assets/fieldreport-page-screenshot.png"),
]
image_path = next((path for path in candidates if path.exists()), None)
if image_path is None:
    searched = ", ".join(str(path) for path in candidates)
    raise FileNotFoundError(f"Screenshot not found. Looked in: {searched}")

class LocalPng:
    def __init__(self, path):
        self.path = Path(path)

    def _repr_png_(self):
        return self.path.read_bytes()

    def __repr__(self):
        return f"LocalPng({self.path})"

page_screenshot = LocalPng(image_path)
page_screenshot
```
